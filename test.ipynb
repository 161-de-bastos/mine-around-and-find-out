```python
from agendasetting.main import *

docs = [
    "Las Bambas incrementó su producción; el Gobierno peruano evaluó nuevos acuerdos con comunidades locales.",
    "La minera MMG y el Ministerio de Energía y Minas coordinaron mejoras en la carretera y en el ferrocarril.",
    "Comuneros de Fuerabamba bloquearon el corredor minero con maquinaria y camiones."
]
```

```python
nlp, mdl = load_models()
txts = preprocess(
    txts = docs,
    nlp = nlp,
    cfg = PREPROCESS_CFG
)
```

```python
from agendasetting.figures import plot_class_distribution

sentiments = run_sentiments(
    txts = docs, 
    mdl = mdl,
    cfg = SENTIMENT_INFERENCE
)

plot_class_distribution(sentiments['avg'])
```

```python
topic, vct, top = run_lda(
    txts = docs, 
    cfg = LDA_CFG
)
```

```python
from agendasetting.metrics import graph_stats, degree_centrality
from agendasetting.figures import plot_degree_hist

spr, ns, pr = run_topic_correlation(
    txts = docs,
    vct = vct,
    mdl = topic,
    graph_cfg = TOPIC_CFG,
    centrality_cfg = PAGERANK_CFG
)

graph_stats(spr)
plot_degree_hist(degree_centrality(spr))
```
